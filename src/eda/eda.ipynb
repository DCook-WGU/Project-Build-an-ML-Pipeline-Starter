1. Fetch the artifact we just created (sample.csv) from W&B and read it with pandas:

```python
import wandb
import pandas as pd
# Note that we use save_code=True in the call to wandb.init so the notebook is uploaded and versioned by W&B
run = wandb.init(project="nyc_airbnb", group="eda", save_code=True)
local_path = wandb.use_artifact("sample.csv:latest").file()
df = pd.read_csv(local_path)
```

```python
# Price column summary
df['price'].describe()
```

```python
Q1 = df['price'].quantile(0.25)
Q3 = df['price'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['price'] < lower) | (df['price'] > upper)]
print(f"Outliers detected: {len(outliers)} ({len(outliers)/len(df)*100:.2f}%)")
print(f"Upper bound for normal range: {upper:.2f}")
```

```python
(df['last_review'].isna() & df['reviews_per_month'].isna()).sum()
```

```python
(df['last_review'].isna() == df['reviews_per_month'].isna()).mean()
```

```python
if (df['last_review'].isna() == df['reviews_per_month'].isna()).all():
    print("All nulls match")
else:
    print("Nulls do not match")
```

```python
zero_reviews = df.query('number_of_reviews == 0')
print(f"The number of rows that contains a total count of zero reviews is {len(zero_reviews)}")
```

```python
if ((df['number_of_reviews'] == 0) == df['last_review'].isna()).all():
    print("All zero review rows match null last_review items")
else:
    print("zero review rows do not match the same last_review null rows")
```

In the "number of reviews" column, the rows that have a value of "0" (zero) match the rows that contain a null/nan value in the 'last review' and 'reviews per month' columns. In otherwords, these three columns have a direct correlation to each other. 

4. Fix some of the little problems we have found in the data with the following code:

```python
# Drop outliers
min_price = 10
max_price = 350
idx = df['price'].between(min_price, max_price)
df = df[idx].copy()
# Convert last_review to datetime
df['last_review'] = pd.to_datetime(df['last_review'])
```

Note how we did not impute missing values. We will do that in the inference pipeline, so we will be able to handle missing values also in production.
